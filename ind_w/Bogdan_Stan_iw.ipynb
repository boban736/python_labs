````python
import math


# Класс Vector
class Vector:
    """Класс для работы с векторами."""

    def __init__(self, coordinates):
        if not coordinates:
            raise ValueError("Вектор не может быть пустым.")
        self.coordinates = list(coordinates)

    def length(self):
        """Возвращает длину (модуль) вектора."""
        return math.sqrt(sum(x ** 2 for x in self.coordinates))

    def dot_product(self, other):
        """Вычисляет скалярное произведение двух векторов."""
        if len(self.coordinates) != len(other.coordinates):
            raise ValueError("Векторы должны иметь одинаковую размерность.")
        return sum(a * b for a, b in zip(self.coordinates, other.coordinates))

    def angle(self, other):
        """Возвращает угол между векторами в градусах."""
        if len(self.coordinates) != len(other.coordinates):
            raise ValueError("Векторы должны иметь одинаковую размерность.")
        cos_a = self.dot_product(other) / (self.length() * other.length())
        # clamp — защита от ε-погрешности при вычислении acos
        cos_a = max(-1.0, min(1.0, cos_a))
        return math.degrees(math.acos(cos_a))

    def add(self, other):
        """Складывает два вектора, возвращает новый."""
        if len(self.coordinates) != len(other.coordinates):
            raise ValueError("Векторы должны иметь одинаковую размерность.")
        return Vector([a + b for a, b in zip(self.coordinates, other.coordinates)])

    def subtract(self, other):
        """Вычитает other из self, возвращает новый вектор."""
        if len(self.coordinates) != len(other.coordinates):
            raise ValueError("Векторы должны иметь одинаковую размерность.")
        return Vector([a - b for a, b in zip(self.coordinates, other.coordinates)])

    def scale(self, scalar):
        """Умножает вектор на скаляр."""
        return Vector([x * scalar for x in self.coordinates])

    def is_collinear(self, other):
        """Проверяет коллинеарность двух векторов."""
        if len(self.coordinates) != len(other.coordinates):
            raise ValueError("Векторы должны иметь одинаковую размерность.")
        # Векторы коллинеарны, если их векторное произведение равно нулю
        # Для произвольной размерности проверяем пропорциональность
        ratios = []
        for a, b in zip(self.coordinates, other.coordinates):
            if b == 0 and a == 0:
                continue
            if b == 0:
                return False
            ratios.append(a / b)
        if not ratios:
            return True
        return all(abs(r - ratios[0]) < 1e-9 for r in ratios)

    def __str__(self):
        return f"Vector{self.coordinates}"


# Класс Matrix
class Matrix:
    """Класс для работы с матрицами."""

    def __init__(self, data):
        if not data or not data[0]:
            raise ValueError("Матрица не может быть пустой.")
        cols = len(data[0])
        for row in data:
            if len(row) != cols:
                raise ValueError("Все строки матрицы должны иметь одинаковую длину.")
        self.data = [list(row) for row in data]
        self.rows = len(self.data)
        self.cols = cols

    def add(self, other):
        """Складывает две матрицы."""
        if self.rows != other.rows or self.cols != other.cols:
            raise ValueError("Матрицы должны иметь одинаковые размеры.")
        result = [
            [self.data[i][j] + other.data[i][j] for j in range(self.cols)]
            for i in range(self.rows)
        ]
        return Matrix(result)

    def subtract(self, other):
        """Вычитает other из self."""
        if self.rows != other.rows or self.cols != other.cols:
            raise ValueError("Матрицы должны иметь одинаковые размеры.")
        result = [
            [self.data[i][j] - other.data[i][j] for j in range(self.cols)]
            for i in range(self.rows)
        ]
        return Matrix(result)

    def multiply_by_vector(self, vector):
        """Умножает матрицу на вектор."""
        if self.cols != len(vector.coordinates):
            raise ValueError("Число столбцов матрицы должно совпадать с размером вектора.")
        result = [
            sum(self.data[i][j] * vector.coordinates[j] for j in range(self.cols))
            for i in range(self.rows)
        ]
        return Vector(result)

    def multiply(self, other):
        """Перемножает матрицы."""
        if self.cols != other.rows:
            raise ValueError("Число столбцов первой матрицы должно равняться числу строк второй.")
        result = [
            [
                sum(self.data[i][k] * other.data[k][j] for k in range(self.cols))
                for j in range(other.cols)
            ]
            for i in range(self.rows)
        ]
        return Matrix(result)

    def transpose(self):
        """Транспонирует матрицу."""
        result = [
            [self.data[i][j] for i in range(self.rows)]
            for j in range(self.cols)
        ]
        return Matrix(result)

    def determinant(self):
        """Считает определитель методом Гаусса."""
        if self.rows != self.cols:
            raise ValueError("Определитель вычисляется только для квадратной матрицы.")
        n = self.rows
        matrix = [row[:] for row in self.data]
        det = 1.0

        for col in range(n):
            # Ищем ненулевой элемент в столбце
            pivot_row = None
            for row in range(col, n):
                if abs(matrix[row][col]) > 1e-9:
                    pivot_row = row
                    break
            if pivot_row is None:
                return 0

            if pivot_row != col:
                matrix[col], matrix[pivot_row] = matrix[pivot_row], matrix[col]
                det *= -1

            det *= matrix[col][col]
            for row in range(col + 1, n):
                factor = matrix[row][col] / matrix[col][col]
                for k in range(col, n):
                    matrix[row][k] -= factor * matrix[col][k]

        return round(det, 9)

    def gaussian_elimination(self, b):
        """Решает Ax = b методом Гаусса. b — список свободных членов."""
        if self.rows != self.cols:
            raise ValueError("Метод Гаусса применяется только для квадратных матриц.")
        if len(b) != self.rows:
            raise ValueError("Размер вектора b должен совпадать с числом строк матрицы.")

        n = self.rows
        # Расширенная матрица [A | b]
        aug = [self.data[i][:] + [b[i]] for i in range(n)]

        for col in range(n):
            # Ищем максимальный элемент (частичный выбор ведущего элемента)
            max_row = max(range(col, n), key=lambda r: abs(aug[r][col]))
            if abs(aug[max_row][col]) < 1e-9:
                raise ValueError("Система не имеет единственного решения.")
            aug[col], aug[max_row] = aug[max_row], aug[col]

            for row in range(col + 1, n):
                factor = aug[row][col] / aug[col][col]
                for k in range(col, n + 1):
                    aug[row][k] -= factor * aug[col][k]

        # Обратный ход
        x = [0.0] * n
        for i in range(n - 1, -1, -1):
            x[i] = (aug[i][n] - sum(aug[i][j] * x[j] for j in range(i + 1, n))) / aug[i][i]

        return [round(val, 9) for val in x]

    def __str__(self):
        rows_str = "\n".join("  " + str(row) for row in self.data)
        return f"Matrix:\n{rows_str}"


# Класс EquationSolver
class EquationSolver:
    """Класс для решения уравнений."""

    def solve_linear(self, a, b):
        """Решает ax + b = 0, возвращает строку с результатом."""
        if a == 0:
            if b == 0:
                return "Уравнение имеет бесконечно много решений."
            else:
                return "Уравнение не имеет решений."
        x = -b / a
        return f"x = {round(x, 9)}"

    def solve_quadratic(self, a, b, c):
        """Решает ax² + bx + c = 0, возвращает строку с результатом."""
        if a == 0:
            return self.solve_linear(b, c)
        discriminant = b ** 2 - 4 * a * c
        if discriminant > 0:
            x1 = (-b + math.sqrt(discriminant)) / (2 * a)
            x2 = (-b - math.sqrt(discriminant)) / (2 * a)
            return f"x₁ = {round(x1, 9)}, x₂ = {round(x2, 9)}"
        elif discriminant == 0:
            x = -b / (2 * a)
            return f"x = {round(x, 9)} (один корень)"
        else:
            real = -b / (2 * a)
            imag = math.sqrt(-discriminant) / (2 * a)
            return f"x₁ = {round(real, 9)} + {round(imag, 9)}i, x₂ = {round(real, 9)} - {round(imag, 9)}i"

    def solve_system(self, matrix, b):
        """Делегирует решение системы методу gaussian_elimination."""
        return matrix.gaussian_elimination(b)

    def display(self, result):
        """Печатает результат."""
        print(f"\nРешение: {result}")


# Вспомогательные функции для ввода данных

def input_vector(prompt):
    """Читает координаты вектора из ввода."""
    while True:
        try:
            raw = input(prompt)
            coords = list(map(float, raw.split()))
            if not coords:
                raise ValueError
            return Vector(coords)
        except ValueError:
            print("Ошибка: введите числа через пробел.")


def input_matrix(prompt):
    """Читает матрицу построчно из ввода."""
    while True:
        try:
            size = input("Введите размер матрицы (строки столбцы): ").split()
            rows, cols = int(size[0]), int(size[1])
            print(prompt)
            data = []
            for i in range(rows):
                while True:
                    try:
                        row = list(map(float, input(f"  Строка {i + 1}: ").split()))
                        if len(row) != cols:
                            raise ValueError
                        data.append(row)
                        break
                    except ValueError:
                        print(f"Ошибка: введите ровно {cols} числа через пробел.")
            return Matrix(data)
        except (ValueError, IndexError):
            print("Ошибка: некорректный ввод размера матрицы.")


def input_float(prompt):
    """Читает вещественное число из ввода."""
    while True:
        try:
            return float(input(prompt))
        except ValueError:
            print("Ошибка: введите число.")


# Интерфейс: работа с векторами

def vectors_menu():
    print("\n--- Работа с векторами ---")
    v1 = input_vector("Введите координаты первого вектора (через пробел): ")

    print("\nВыберите операцию:")
    print("  1. Длина вектора")
    print("  2. Скалярное произведение")
    print("  3. Угол между векторами")
    print("  4. Сложение векторов")
    print("  5. Вычитание векторов")
    print("  6. Умножение на скаляр")
    print("  7. Проверка коллинеарности")

    choice = input("Ваш выбор: ").strip()

    try:
        if choice == "1":
            print(f"\nДлина вектора: {round(v1.length(), 6)}")

        elif choice == "2":
            v2 = input_vector("Введите координаты второго вектора: ")
            print(f"\nСкалярное произведение: {v1.dot_product(v2)}")

        elif choice == "3":
            v2 = input_vector("Введите координаты второго вектора: ")
            print(f"\nУгол между векторами: {round(v1.angle(v2), 6)}°")

        elif choice == "4":
            v2 = input_vector("Введите координаты второго вектора: ")
            result = v1.add(v2)
            print(f"\nРезультат сложения: {result}")

        elif choice == "5":
            v2 = input_vector("Введите координаты второго вектора: ")
            result = v1.subtract(v2)
            print(f"\nРезультат вычитания: {result}")

        elif choice == "6":
            scalar = input_float("Введите скаляр: ")
            result = v1.scale(scalar)
            print(f"\nРезультат умножения: {result}")

        elif choice == "7":
            v2 = input_vector("Введите координаты второго вектора: ")
            if v1.is_collinear(v2):
                print("\nВекторы коллинеарны.")
            else:
                print("\nВекторы не коллинеарны.")

        else:
            print("Неверный выбор.")

    except ValueError as e:
        print(f"\nОшибка: {e}")


# Интерфейс: работа с матрицами

def matrices_menu():
    print("\n--- Работа с матрицами ---")

    print("\nВыберите операцию:")
    print("  1. Сложение матриц")
    print("  2. Вычитание матриц")
    print("  3. Умножение матрицы на вектор")
    print("  4. Умножение матриц")
    print("  5. Транспонирование матрицы")
    print("  6. Определитель матрицы")
    print("  7. Решение системы линейных уравнений (метод Гаусса)")

    choice = input("Ваш выбор: ").strip()

    try:
        if choice in ("1", "2"):
            m1 = input_matrix("Введите элементы первой матрицы:")
            m2 = input_matrix("Введите элементы второй матрицы:")
            result = m1.add(m2) if choice == "1" else m1.subtract(m2)
            print(f"\nРезультат:\n{result}")

        elif choice == "3":
            m = input_matrix("Введите элементы матрицы:")
            v = input_vector("Введите координаты вектора: ")
            result = m.multiply_by_vector(v)
            print(f"\nРезультат: {result}")

        elif choice == "4":
            m1 = input_matrix("Введите элементы первой матрицы:")
            m2 = input_matrix("Введите элементы второй матрицы:")
            result = m1.multiply(m2)
            print(f"\nРезультат:\n{result}")

        elif choice == "5":
            m = input_matrix("Введите элементы матрицы:")
            result = m.transpose()
            print(f"\nТранспонированная матрица:\n{result}")

        elif choice == "6":
            m = input_matrix("Введите элементы квадратной матрицы:")
            det = m.determinant()
            print(f"\nОпределитель: {det}")

        elif choice == "7":
            m = input_matrix("Введите коэффициенты системы (матрица A):")
            print("Введите свободные члены (вектор b):")
            b = list(map(float, input("  b: ").split()))
            solution = m.gaussian_elimination(b)
            print("\nРешение системы:")
            for i, x in enumerate(solution):
                print(f"  x{i + 1} = {x}")

        else:
            print("Неверный выбор.")

    except ValueError as e:
        print(f"\nОшибка: {e}")


# Интерфейс: решение уравнений

def equations_menu():
    print("\n--- Решение уравнений ---")
    print("Выберите тип уравнения:")
    print("  1. Линейное уравнение (ax + b = 0)")
    print("  2. Квадратное уравнение (ax² + bx + c = 0)")
    print("  3. Система линейных уравнений")

    choice = input("Ваш выбор: ").strip()
    solver = EquationSolver()

    try:
        if choice == "1":
            print("Введите коэффициенты a и b:")
            a = input_float("  a = ")
            b = input_float("  b = ")
            solver.display(solver.solve_linear(a, b))

        elif choice == "2":
            print("Введите коэффициенты a, b, c:")
            a = input_float("  a = ")
            b = input_float("  b = ")
            c = input_float("  c = ")
            solver.display(solver.solve_quadratic(a, b, c))

        elif choice == "3":
            m = input_matrix("Введите коэффициенты системы (матрица A):")
            print("Введите свободные члены (вектор b):")
            b = list(map(float, input("  b: ").split()))
            solution = solver.solve_system(m, b)
            print("\nРешение системы:")
            for i, x in enumerate(solution):
                print(f"  x{i + 1} = {x}")

        else:
            print("Неверный выбор.")

    except ValueError as e:
        print(f"\nОшибка: {e}")


# Главное меню

def main():
    while True:
        print("\n=============================")
        print("   Математическая библиотека")
        print("=============================")
        print("1. Работа с векторами")
        print("2. Работа с матрицами")
        print("3. Решение уравнений")
        print("0. Выход")

        choice = input("\nВыберите действие: ").strip()

        if choice == "1":
            vectors_menu()
        elif choice == "2":
            matrices_menu()
        elif choice == "3":
            equations_menu()
        elif choice == "0":
            print("До свидания!")
            break
        else:
            print("Неверный выбор. Попробуйте снова.")


main()
````
